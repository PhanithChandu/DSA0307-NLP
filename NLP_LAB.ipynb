#EXP 1 - Write program demonstrates how to use regular expressions in Python to match and search for patterns in text.

```python
import re

text = "My email is student123@gmail.com and my phone number is 9876543210."

email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
email = re.search(email_pattern, text)

if email:
    print("Email found:", email.group())
else:
    print("Email not found.")

match = re.match(r"My", text)

if match:
    print("Matched at the beginning:", match.group())
else:
    print("No match at the beginning.")

phone_pattern = r'\b\d{10}\b'
phone = re.search(phone_pattern, text)

if phone:
    print("Phone number found:", phone.group())
else:
    print("Phone number not found.")
```

# EXP 2 - Implement a basic finite state automaton that recognizes a specific language or pattern. In this example, we'll create a simple automaton to match strings ending with 'ab' using python.



```python
def fsa(string):
    state = 0

    for ch in string:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 1:
            if ch == 'a':
                state = 1
            elif ch == 'b':
                state = 2
            else:
                state = 0
        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0

    if state == 2:
        print("Accepted")
    else:
        print("Rejected")

s = input("Enter a string: ")
fsa(s)
```

simple machine to generate plural forms of English nouns using python.


```python
def pluralize(noun):
    if noun.endswith(("s", "x", "z", "ch", "sh")):
        return noun + "es"
    elif noun.endswith("y") and noun[-2] not in "aeiou":
        return noun[:-1] + "ies"
    else:
        return noun + "s"

word = input("Enter a noun: ")
print("Plural form:", pluralize(word))
```

#EXP 06 - Implement a basic N-gram model for text generation. For example, generate text using a bigram model using python.

```python
import random

text = "the cat sat on the mat the cat ate the fish"

words = text.split()

bigrams = {}

for i in range(len(words) - 1):
    key = words[i]
    value = words[i + 1]
    if key not in bigrams:
        bigrams[key] = []
    bigrams[key].append(value)

current = random.choice(words)
generated = [current]

for _ in range(9):
    if current in bigrams:
        current = random.choice(bigrams[current])
        generated.append(current)
    else:
        break

print("Generated Text:")
print(" ".join(generated))
```

#EXP 10 - Implement transformation-based tagging using a set of transformation rules, apply a simple rule to tag words using python.

```python
sentence = "The cat is running quickly"

words = sentence.split()

tags = []

for word in words:
    if word.lower() in ["the", "a", "an"]:
        tags.append((word, "DT"))
    elif word.lower() in ["is", "am", "are", "was", "were"]:
        tags.append((word, "VB"))
    else:
        tags.append((word, "NN"))

transformed_tags = []

for word, tag in tags:
    if word.endswith("ing"):
        tag = "VBG"
    elif word.endswith("ly"):
        tag = "RB"
    transformed_tags.append((word, tag))

print("Tagged Sentence:")
for word, tag in transformed_tags:
    print(word, ":", tag)
```

#EXP 11 - Implement a simple top-down parser for context-free grammars using python.

```python
def parse(grammar, sentence):
    def expand(symbol, current_sentence, index):
        if index == len(sentence):
            return symbol == []

        if not symbol:
            return False

        current_symbol = symbol[0]

        # If it's a terminal, match directly
        if current_symbol in [word for rule in grammar for word in rule[1] if not word.isupper()]:
            if sentence[index] == current_symbol:
                return expand(symbol[1:], current_sentence, index + 1)
            else:
                return False
        # If it's a non-terminal, try all its productions
        else:
            for rule in grammar:
                if rule[0] == current_symbol:
                    if expand(rule[1] + symbol[1:], current_sentence, index):
                        return True
            return False

    # Start parsing with the start symbol (assuming the first rule defines it)
    start_symbol = grammar[0][0]
    return expand([start_symbol], sentence, 0)

# Define a simple context-free grammar (Chomsky Normal Form for simplicity)
# Rules are represented as (non_terminal, [production_symbols])
grammar = [
    ('S', ['NP', 'VP']),
    ('NP', ['Det', 'N']),
    ('VP', ['V', 'NP']),
    ('Det', ['the']),
    ('N', ['cat']),
    ('N', ['dog']),
    ('V', ['chases'])
]

# Example usage
sentence1 = ['the', 'cat', 'chases', 'the', 'dog']
sentence2 = ['the', 'dog', 'eats', 'the', 'cat'] # Should be rejected

print(f"Sentence 1: {sentence1}")
if parse(grammar, sentence1):
    print("Accepted")
else:
    print("Rejected")

print(f"\nSentence 2: {sentence2}")
if parse(grammar, sentence2):
    print("Accepted")
else:
    print("Rejected")
```

#EXP 12 - 12. Implement an Earley parser for context-free grammars using a simple python program.

```python
class EarleyParser:
    def __init__(self, grammar):
        self.grammar = grammar
        self.chart = []

    def parse(self, words):
        self.chart = [[] for _ in range(len(words) + 1)]

        # Add initial state for the start symbol
        self.predict(self.grammar['S'][0][0], 0, 0)

        for i in range(len(words) + 1):
            for state in list(self.chart[i]): # Iterate over a copy as chart[i] might be modified
                if state['dot'] < len(state['production']):
                    # Current symbol is a non-terminal
                    if state['production'][state['dot']].isupper():
                        self.predict(state['production'][state['dot']], i, i)
                    # Current symbol is a terminal
                    else:
                        self.scan(state, i, words)
                else:
                    self.complete(state, i)

        # Check if the sentence is accepted
        for state in self.chart[len(words)]:
            if state['name'] == self.grammar['S'][0][0] and state['dot'] == len(state['production']) and state['start'] == 0:
                return True
        return False

    def predict(self, non_terminal, current_position, start_position):
        for production in self.grammar.get(non_terminal, []):
            new_state = {
                'name': non_terminal,
                'production': production,
                'dot': 0,
                'start': start_position
            }
            if new_state not in self.chart[current_position]:
                self.chart[current_position].append(new_state)

    def scan(self, state, current_position, words):
        if current_position < len(words):
            next_symbol = state['production'][state['dot']]
            if next_symbol == words[current_position]:
                new_state = state.copy()
                new_state['dot'] += 1
                if new_state not in self.chart[current_position + 1]:
                    self.chart[current_position + 1].append(new_state)

    def complete(self, state, current_position):
        for old_state in self.chart[state['start']]:
            if old_state['dot'] < len(old_state['production']) and \
               old_state['production'][old_state['dot']] == state['name']:
                new_state = old_state.copy()
                new_state['dot'] += 1
                if new_state not in self.chart[current_position]:
                    self.chart[current_position].append(new_state)

# Define a context-free grammar in Chomsky Normal Form (or similar for simplicity)
# Productions are defined as a dictionary where keys are non-terminals
grammar = {
    'S': [['NP', 'VP']],
    'NP': [['Det', 'N'], ['N']],
    'VP': [['V', 'NP'], ['V']],
    'Det': ['the'],
    'N': ['cat', 'dog', 'fish'],
    'V': ['chases', 'eats']
}

parser = EarleyParser(grammar)

# Test sentences
sentence1 = ['the', 'cat', 'chases', 'the', 'dog']
sentence2 = ['the', 'dog', 'eats', 'fish']
sentence3 = ['cat', 'eats']
sentence4 = ['the', 'cat', 'eats', 'the'] # Should be rejected based on grammar

print(f"Parsing: {sentence1}")
print(f"Accepted: {parser.parse(sentence1)}\n")

print(f"Parsing: {sentence2}")
print(f"Accepted: {parser.parse(sentence2)}\n")

print(f"Parsing: {sentence3}")
print(f"Accepted: {parser.parse(sentence3)}\n")

print(f"Parsing: {sentence4}")
print(f"Accepted: {parser.parse(sentence4)}\n")
```

EXP 13 - Generate a parse tree for a given sentence using a context-free grammar using python program.

```python
import nltk
from nltk import CFG
from nltk.parse import ChartParser

# Define CFG
grammar = CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP

Det -> 'the' | 'a'
N -> 'cat' | 'dog'
V -> 'chased' | 'saw'
""")

# Create parser
parser = ChartParser(grammar)

# Input sentence
sentence = "the cat chased a dog".split()

# Generate parse tree
for tree in parser.parse(sentence):
    print("Parse Tree:")
    print(tree)
    tree.pretty_print()
```

#EXP 14 - Create a program in python to check for agreement in sentences based on a context-free grammar's rules.

```python
class Parser:
    def __init__(self):
        self.g = {
            'S': [['NP_SG', 'VP_SG'], ['NP_PL', 'VP_PL']],
            'NP_SG': [['the', 'cat'], ['a', 'dog']],
            'NP_PL': [['the', 'cats'], ['the', 'dogs']],
            'VP_SG': [['runs'], ['eats']],
            'VP_PL': [['run'], ['eat']]
        }

    def parse(self, s):
        return self.check('S', s.split())

    def check(self, sym, tokens):
        if sym not in self.g:
            return [tokens[1:]] if tokens and tokens[0] == sym else []

        result = []
        for rule in self.g[sym]:
            rem = [tokens]
            for item in rule:
                temp = []
                for r in rem:
                    temp += self.check(item, r)
                rem = temp
            result += rem
        return result

p = Parser()

tests = [
    "the cat runs",
    "the cats run",
    "the cat run",
    "the dogs eats",
    "a dog eats"
]

for t in tests:
    print(f"{t} -->", "Accepted" if [] in p.parse(t) else "Rejected")
```

#EXP 15 - Implement probabilistic context-free grammar parsing for a sentence using python.

```python
import nltk
from nltk import PCFG
from nltk.parse import ViterbiParser

# Define Probabilistic CFG
grammar = PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.6] | 'John' [0.4]
VP -> V NP [1.0]

Det -> 'the' [0.5] | 'a' [0.5]
N -> 'cat' [0.5] | 'dog' [0.5]
V -> 'chased' [0.6] | 'saw' [0.4]
""")

# Create parser
parser = ViterbiParser(grammar)

# Input sentence
sentence = "the cat chased a dog".split()

# Parse sentence
for tree in parser.parse(sentence):
    print(tree)
    print("Probability:", tree.prob())
```

#EXP 18 - Implement a simple FOPC parser for basic logical expressions using python program.

```python
import re

# Parse FOPC expression
def parse(expr):
    pattern = r'([A-Za-z]+)\((.*)\)'
    match = re.match(pattern, expr)

    if match:
        predicate = match.group(1)
        arguments = [arg.strip() for arg in match.group(2).split(",")]

        print("Predicate :", predicate)
        print("Arguments :", arguments)
    else:
        print("Invalid FOPC Expression")

# Input expression
expression = "Loves(John,Mary)"

parse(expression)
```

#EXP 20 - Implement a basic information retrieval system using TF-IDF (Term Frequency-Inverse Document Frequency) for document ranking using python.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Documents
docs = [
    "Python is a programming language",
    "Machine learning uses Python",
    "Information retrieval uses TF IDF"
]

# User query
query = ["Python for machine learning"]

# TF-IDF Vectorization
vectorizer = TfidfVectorizer()
tfidf = vectorizer.fit_transform(docs + query)

# Calculate similarity
scores = cosine_similarity(tfidf[-1], tfidf[:-1])

# Display ranked documents
ranking = sorted(enumerate(scores[0]), key=lambda x: x[1], reverse=True)

print("Document Ranking:")
for i, score in ranking:
    print(f"Doc {i+1}: {docs[i]}  (Score: {score:.2f})")
```
